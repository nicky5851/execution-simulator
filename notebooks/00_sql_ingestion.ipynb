```python
import duckdb
con = duckdb.connect()
con.execute("""
CREATE OR REPLACE TABLE btc_1sec_core AS
SELECT
        system_time,
        midpoint,
        spread,
        buys,
        sells
FROM read_csv_auto('../data/raw/BTC_1sec.csv')
    
""")

```

```python
con.execute("""
COPY btc_1sec_core
TO '../data/processed/btc_1sec_core.csv'
(HEADER, DELIMITER ',');

""")
```
